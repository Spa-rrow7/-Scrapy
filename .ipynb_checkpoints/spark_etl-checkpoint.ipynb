```python
import logging
import os
import sys
from pathlib import Path
from urllib.parse import urlparse

sys.path = [p for p in sys.path if "C:\\Spark" not in p and "C:/Spark" not in p]
os.environ.pop("SPARK_HOME", None)
os.environ.pop("HADOOP_HOME", None)
os.environ.pop("PYTHONPATH", None)
os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable
from dotenv import load_dotenv
from pyspark.sql import SparkSession, Window
from pyspark.sql import functions as F
from pyspark.sql.types import (
    ArrayType, BooleanType, DecimalType, IntegerType,
    MapType, StringType, StructField, StructType, TimestampType
)
```

```python
BASE_DIR = Path.cwd()
LOGS_DIR = BASE_DIR / "logs"
LOGS_DIR.mkdir(parents=True, exist_ok=True)

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] spark_etl: %(message)s",
    handlers=[
        logging.FileHandler(LOGS_DIR / "spark_etl.log", mode="a", encoding="utf-8"),
        logging.StreamHandler(sys.stdout),
    ],
    force=True,
)
logger = logging.getLogger("spark_etl")

load_dotenv(BASE_DIR / ".env")
DATABASE_URL = os.getenv("DATABASE_URL")
JDBC_JAR_PATH = os.getenv("POSTGRES_JDBC_JAR", "")
```

```python
parsed_db = urlparse(DATABASE_URL.replace("postgresql+psycopg2://", "postgresql://"))
JDBC_URL = f"jdbc:postgresql://{parsed_db.hostname}:{parsed_db.port or 5432}{parsed_db.path}"
DB_PROPERTIES = {
    "user": parsed_db.username,
    "password": parsed_db.password,
    "driver": "org.postgresql.Driver",
    "stringtype": "unspecified",
}

builder = (
    SparkSession.builder
    .appName("Skifmusic_ETL_Pipeline")
    .master("local[*]")
    .config("spark.sql.session.timeZone", "UTC")
)
builder = (
    builder.config("spark.jars", JDBC_JAR_PATH)
    if JDBC_JAR_PATH
    else builder.config("spark.jars.packages", "org.postgresql:postgresql:42.7.3")
)

spark = builder.getOrCreate()
spark.sparkContext.setLogLevel("WARN")
logger.info("SparkSession запущена! Подключение: %s", JDBC_URL)
```

```python
raw_df = spark.read.jdbc(url=JDBC_URL, table="raw_products", properties=DB_PROPERTIES)
total_raw = raw_df.count()

logger.info("Загружено строк из сырого слоя raw_products: %d", total_raw)
raw_df.select("id", "url", "parsed_at", "created_at").show(5, truncate=50)
```

```python
review_schema = ArrayType(StructType([
    StructField("review_id", StringType(), True),
    StructField("author", StringType(), True),
    StructField("rating", IntegerType(), True),
    StructField("title", StringType(), True),
    StructField("text", StringType(), True),
    StructField("verified", BooleanType(), True),
    StructField("date", StringType(), True),
    StructField("helpful", IntegerType(), True),
]))

payload_schema = StructType([
    StructField("url", StringType(), True),
    StructField("sku", StringType(), True),
    StructField("title", StringType(), True),
    StructField("description", StringType(), True),
    StructField("category", ArrayType(StringType()), True),
    StructField("brand", StringType(), True),
    StructField("price", StringType(), True),
    StructField("rating", StringType(), True),
    StructField("reviews_count", IntegerType(), True),
    StructField("seller", StringType(), True),
    StructField("condition", StringType(), True),
    StructField("availability", StringType(), True),
    StructField("characteristics", MapType(StringType(), StringType()), True),
    StructField("comments", review_schema, True),
    StructField("image_urls", ArrayType(StringType()), True),
    StructField("parsed_at", StringType(), True),
])

parsed_df = raw_df.withColumn("data", F.from_json(F.col("raw_data"), payload_schema))

filtered_df = parsed_df.filter(
    F.col("url").isNotNull() & (F.trim(F.col("url")) != "") &
    F.col("data.title").isNotNull() & (F.trim(F.col("data.title")) != "") &
    F.col("data.price").isNotNull()
)
logger.info("После фильтрации валидных записей: %d (отброшено: %d)", filtered_df.count(), total_raw - filtered_df.count())

window_spec = Window.partitionBy("url").orderBy(F.col("parsed_at").desc())
dedup_df = (
    filtered_df
    .withColumn("rn", F.row_number().over(window_spec))
    .filter(F.col("rn") == 1)
    .drop("rn")
)
logger.info("После дедупликации по URL уникальных товаров: %d", dedup_df.count())
```

```python
cleaned_products_df = dedup_df.select(
    F.trim(F.col("url")).alias("url"),
    F.coalesce(F.nullif(F.trim(F.col("data.sku")), ""), F.regexp_extract(F.col("url"), r"/product/(\d+)-", 1), F.lit("UNKNOWN")).alias("sku"),
    F.trim(F.col("data.title")).alias("title"),
    F.coalesce(F.nullif(F.trim(F.col("data.description")), ""), F.trim(F.col("data.title"))).alias("description"),
    F.when(F.col("data.category").isNotNull() & (F.size(F.col("data.category")) > 0), F.col("data.category")).otherwise(F.array(F.lit("Гитары"))).alias("category"),
    F.coalesce(F.nullif(F.trim(F.col("data.brand")), ""), F.col("data.characteristics")["Производитель"], F.lit("Не указан")).alias("brand"),
    F.col("data.price").cast(DecimalType(12, 2)).alias("price"),
    F.coalesce(F.col("data.rating").cast(DecimalType(3, 2)), F.lit(0.00).cast(DecimalType(3, 2))).alias("rating"),
    F.coalesce(F.col("data.reviews_count"), F.lit(0)).cast(IntegerType()).alias("reviews_count"),
    F.coalesce(F.nullif(F.trim(F.col("data.seller")), ""), F.lit("SKIFMUSIC")).alias("seller"),
    F.coalesce(F.nullif(F.trim(F.col("data.condition")), ""), F.lit("Новое")).alias("condition"),
    F.coalesce(F.nullif(F.trim(F.col("data.availability")), ""), F.lit("In Stock")).alias("availability"),
    F.to_json(F.coalesce(F.col("data.characteristics"), F.create_map())).alias("characteristics"),
    F.coalesce(F.col("data.image_urls"), F.array().cast(ArrayType(StringType()))).alias("image_urls"),
    F.col("parsed_at").cast(TimestampType()).alias("parsed_at"),
    F.current_timestamp().alias("created_at"),
    F.col("data.characteristics").alias("_raw_chars"),
    F.col("data.comments").alias("_raw_comments"),
).filter(F.col("price").isNotNull() & (F.col("price") > 0))

cleaned_products_df.drop("_raw_chars", "_raw_comments", "characteristics", "description").printSchema()
cleaned_products_df.select("sku", "title", "brand", "price", "rating", "availability").show(5, truncate=30)
```
